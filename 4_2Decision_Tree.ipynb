```python
#defining input and output

x=df.iloc[:-1]
y=df.iloc[:,-1]
print("Shape of x:",x.shape)
print("Shape of y:",y.shape)
```

```python
x = df.iloc[:, :-1].copy()
y = df.iloc[:, -1].copy()

print("Shape of x:", x.shape)
print("Shape of y:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.20, random_state=51
)

print("Shape of x_train:", x_train.shape)
print("Shape of x_test:", x_test.shape)
print("Shape of y_train:", y_train.shape)
print("Shape of y_test:", y_test.shape)
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.20, random_state=51
)

print("Shape of x_train:", x_train.shape)
print("Shape of x_test:", x_test.shape)
print("Shape of y_train:", y_train.shape)
print("Shape of y_test:", y_test.shape)
```

```python
from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier(random_state=51)

dt.fit(x_train, y_train)

print("Model trained successfully")
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dc=DecisionTreeClassifier(criterion='gini',max_depth=10)
dc.fit(x_train,y_train)
```

```python
dc.score(x_test,y_test)
```

```python
#Random Forest Classifier

from sklearn.ensemble import RandomForestClassifier
```

```python
clr=RandomForestClassifier(n_estimators=100,criterion='gini')
clr.fit(x_train,y_train)
```

```python
clr.score(x_test,y_test)
```

```python
#Evolution Matrix classification Model

from sklearn.metrics import confusion_matrix
```

```python
y_pred = clr.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

```python
a = accuracy_score(y_test, y_pred)
print("Accuracy:", a)
```

```python
p = precision_score(y_test, y_pred)
print("Precision:", p)
```

```python
r = recall_score(y_test, y_pred)
print("Recall:", r)
```

```python
f = f1_score(y_test, y_pred)
print("F1 Score:", f)
```
