```python
# Import Libraries
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Load Dataset
df = pd.read_csv('/content/house price prediction (1).csv')

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

print("\nFirst 5 rows:")
display(df.head())

# Check Dataset
print("\nDataset Information:")
print(df.info())

print("\nMissing Values:")
print(df.isnull().sum().sum())

# Separate Features and Target
x = df.drop('price', axis=1)
y = df['price']

print("\nShape of X:", x.shape)
print("Shape of Y:", y.shape)

# Train Test Split
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.20, random_state=51
)

print("\nTrain-Test Split:")
print("X Train:", x_train.shape)
print("X Test:", x_test.shape)
print("Y Train:", y_train.shape)
print("Y Test:", y_test.shape)

# Simple Linear Regression
x_simple = df.iloc[:, [0]]
y_simple = df['price']

x_train_s, x_test_s, y_train_s, y_test_s = train_test_split(
    x_simple, y_simple, test_size=0.20, random_state=51
)

slr = LinearRegression()
slr.fit(x_train_s, y_train_s)

y_pred_slr = slr.predict(x_test_s)

print("\nSimple Linear Regression")
print("MAE:", mean_absolute_error(y_test_s, y_pred_slr))
print("MSE:", mean_squared_error(y_test_s, y_pred_slr))
print("R2 Score:", r2_score(y_test_s, y_pred_slr))

# Multiple Linear Regression
mlr = LinearRegression()
mlr.fit(x_train, y_train)

y_pred_mlr = mlr.predict(x_test)

print("\nMultiple Linear Regression")
print("MAE:", mean_absolute_error(y_test, y_pred_mlr))
print("MSE:", mean_squared_error(y_test, y_pred_mlr))
print("R2 Score:", r2_score(y_test, y_pred_mlr))

# Decision Tree Regression
dt_reg = DecisionTreeRegressor(random_state=51)
dt_reg.fit(x_train, y_train)

y_pred_dt = dt_reg.predict(x_test)

print("\nDecision Tree Regression")
print("MAE:", mean_absolute_error(y_test, y_pred_dt))
print("MSE:", mean_squared_error(y_test, y_pred_dt))
print("R2 Score:", r2_score(y_test, y_pred_dt))

# Random Forest Regression
rf_reg = RandomForestRegressor(
    n_estimators=100,
    random_state=51
)

rf_reg.fit(x_train, y_train)

y_pred_rf = rf_reg.predict(x_test)

print("\nRandom Forest Regression")
print("MAE:", mean_absolute_error(y_test, y_pred_rf))
print("MSE:", mean_squared_error(y_test, y_pred_rf))
print("R2 Score:", r2_score(y_test, y_pred_rf))

# KNN Regression
knn_reg = KNeighborsRegressor(n_neighbors=3)
knn_reg.fit(x_train, y_train)

y_pred_knn = knn_reg.predict(x_test)

print("\nKNN Regression")
print("MAE:", mean_absolute_error(y_test, y_pred_knn))
print("MSE:", mean_squared_error(y_test, y_pred_knn))
print("R2 Score:", r2_score(y_test, y_pred_knn))

# Final Model Comparison
results = pd.DataFrame({
    'Model': [
        'Simple Linear Regression',
        'Multiple Linear Regression',
        'Decision Tree Regression',
        'Random Forest Regression',
        'KNN Regression'
    ],
    'MAE': [
        mean_absolute_error(y_test_s, y_pred_slr),
        mean_absolute_error(y_test, y_pred_mlr),
        mean_absolute_error(y_test, y_pred_dt),
        mean_absolute_error(y_test, y_pred_rf),
        mean_absolute_error(y_test, y_pred_knn)
    ],
    'MSE': [
        mean_squared_error(y_test_s, y_pred_slr),
        mean_squared_error(y_test, y_pred_mlr),
        mean_squared_error(y_test, y_pred_dt),
        mean_squared_error(y_test, y_pred_rf),
        mean_squared_error(y_test, y_pred_knn)
    ],
    'R2 Score': [
        r2_score(y_test_s, y_pred_slr),
        r2_score(y_test, y_pred_mlr),
        r2_score(y_test, y_pred_dt),
        r2_score(y_test, y_pred_rf),
        r2_score(y_test, y_pred_knn)
    ]
})

display(results)
```
